```python

for col in ['Age' , 'RestingBP','Cholesterol','MaxHR']:
    plt.figure()     
    sns.histplot(data=df, x=col,kde=True , palette="rainbow")
    plt.title(f"Distribution of {col}")  # Clearly label each plot
    plt.show()  # Display the plot immediately
```

```python
mean_val = int(df[df['Cholesterol'] != 0]['Cholesterol'].mean()) 
df.loc[df['Cholesterol'] == 0, 'Cholesterol'] = mean_val
```

```python
mean_val = int(df[df['RestingBP'] != 0]['RestingBP'].mean()) 
df.loc[df['RestingBP'] == 0, 'RestingBP'] = mean_val
```
