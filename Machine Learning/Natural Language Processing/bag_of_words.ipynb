```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
documents = [
    "I like pizza",
    "Pizza is best",
    "I love pasta",
    "Pasta is great"
]
```

```python
vectorizer = CountVectorizer()
```

```python
X=vectorizer.fit_transform(documents)
```

```python
vectorizer.get_feature_names_out()
```

```python
X.toarray()
```

### Pros

1. Easy to implenent
2. Fixed size 
3. Work with ML models
4. Fast and Efficient

### Cons

1. Sparse Matrix  ( too many 0s)
2. Out of vocabalury
3. Semantic Meaning (Better than one hot encoding)
4. Out of Order
