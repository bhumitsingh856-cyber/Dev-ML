```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import string
```

```python
df = pd.read_csv("../Supervised Learning/Data Sets/emotions.csv",header=None,names=['text','emotion'])
```

```python
df.isna().sum()
```

```python
unique_emotions=df['emotion'].unique()
```

```python
d={}
i=0
for j in unique_emotions:
    d[j] = i
    i+=1
```

```python
df['emotion']=df['emotion'].map(d)
```

```python
df.head()
```

```python
# Convert to lower case 
df['text'] = df['text'].apply(lambda x:x.lower())
```

```python
# Remove punctuations

def rm_pn(x):
    return x.translate(str.maketrans('','',string.punctuation))

df['text'] = df['text'].apply(rm_pn)
```

```python
# Remove numbers 

def rm_num(x):
    txt=""
    for i in x:
        if(not i.isdigit()):
            txt+=i
    return txt

df['text'] = df['text'].apply(rm_num)
```

```python
# Remove emojies

def rm_emo(x):
    txt=""
    for i in x:
        if(not i.isascii()):
            txt+=i
    return txt

df['text'] = df['text'].apply(rm_emo)
```

```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
nltk.download('punkt_tab')
nltk.download('stopwords')
```

```python
stop_words = set(stopwords.words('english'))
```

```python
# Tokenizing

def rm_stop(x):
    arr = word_tokenize(x)
    cleaned=[]
    for i in arr:
        if(not i in stop_words):
            cleaned.append(i)
    return " ".join(cleaned)
```

```python
df['text']=df['text'].apply(rm_stop)
```
