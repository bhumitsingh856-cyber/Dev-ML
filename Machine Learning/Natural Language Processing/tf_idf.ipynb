```python
#TF-IDF : Term frequency Inverse Document Frequency

# TF= ( No. of Occurance of Term in Frequency ) / ( Total no. of Terms in Document )

# IDF = loge( Total no. of doc in carpus ) / ( Total no. of Doc in Term )
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
documents = [
    "I like pizza",
    "Pizza is best",
    "I love pasta",
    "Pasta is great"
]
```

```python
vectorizer = TfidfVectorizer() 
```

```python
X=vectorizer.fit_transform(documents)
```
